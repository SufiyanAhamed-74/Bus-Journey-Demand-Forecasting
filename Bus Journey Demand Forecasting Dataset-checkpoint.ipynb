# Bus Journey Demand Forecasting Dataset

### Dataset Features
#### File: transactions.csv
This file contains detailed transaction-level booking and search history for each bus route and journey date.

Columns
- doj : Date of journey
- doi : Date of issue (booking date)
- dbd : Days before departure from the booking date
- srcid : Source city ID
- destid : Destination city ID
- srcid_region : Region (state) of the source city
- destid_region : Region (state) of the destination city
- srcid_tier : Tier classification of the source city
- destid_tier : Tier classification of the destination city
- cumsum_seatcount : Cumulative number of seats booked up to that date
- cumsum_searchcount : Cumulative number of searches up to that date
#### File: train.csv

This file contains the training dataset with historical journey-level booking information.

Columns:
- doj – Date of journey
- srcid – Source city ID
- destid – Destination city ID
- final_seatcount – Total number of seats booked for the journey (target variable)

#### File: test.csv

This file contains the test dataset used for generating demand forecasts.

Columns:
- route_key – Unique identifier for each route entry
- doj – Date of journey
- srcid – Source city ID
- destid – Destination city ID

#### File: sample_submission.csv

This file provides the required format for submitting model predictions.

Columns:
- route_key – Unique identifier for each test route
- final_seatcount – Predicted total number of seats booked for the journey

### Outlier Detection

```python
outliers = []
for col in numeric_cols:
    Q1 = train[col].quantile(0.25)
    Q3 = train[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    count = ((train[col] < lower_bound) | (train[col] > upper_bound)).sum()
    
    outliers.append({
    'column':col,
    'count':count
    })
    
```

```python
print(train_model.loc[train_indices, "doj"].min())
print(train_model.loc[train_indices, "doj"].max())
```

```python
from sklearn.model_selection import TimeSeriesSplit

tscv = TimeSeriesSplit(n_splits=3)
```

```python
dates_sorted = train_model.loc[train_indices, "doj"].reset_index(drop=True)

X_train_sorted = X_train_sorted.reset_index(drop=True)
y_train_sorted = y_train_sorted.reset_index(drop=True)
```

```python
unique_dates = dates_sorted.sort_values().unique()

n_dates = len(unique_dates)

fold_size = n_dates // 4

date_folds = []

for i in range(1, 4):
    
    train_end = fold_size * i
    val_end = fold_size * (i + 1)

    train_dates = unique_dates[:train_end]
    val_dates = unique_dates[train_end:val_end]

    train_idx = dates_sorted.isin(train_dates).to_numpy().nonzero()[0]
    val_idx = dates_sorted.isin(val_dates).to_numpy().nonzero()[0]

    date_folds.append((train_idx, val_idx))
```

```python
for i, (tr_idx, va_idx) in enumerate(date_folds, 1):

    print(f"\nFold {i}")

    print(
        "Train:",
        dates_sorted.iloc[tr_idx].min(),
        "to",
        dates_sorted.iloc[tr_idx].max()
    )

    print(
        "Validation:",
        dates_sorted.iloc[va_idx].min(),
        "to",
        dates_sorted.iloc[va_idx].max()
    )

    print("Train rows:", len(tr_idx))
    print("Validation rows:", len(va_idx))
```

```python
param_dist = {
    "n_estimators": [100, 200, 300],
    "max_depth": [10, 15, 20, None],
    "min_samples_split": [2, 5, 10],
    "min_samples_leaf": [1, 2, 4],
    "max_features": ["sqrt", 0.7, 1.0]
}
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import RandomizedSearchCV

rf_base = RandomForestRegressor(
    random_state=42,
    n_jobs=-1
)

rf_search = RandomizedSearchCV(
    estimator=rf_base,
    param_distributions=param_dist,
    n_iter=15,
    scoring="neg_mean_absolute_error",
    cv=date_folds,
    random_state=42,
    n_jobs=-1,
    verbose=1
)
```

```python
rf_search.fit(X_train_sorted, y_train_sorted)
```

```python
print("Best Parameters:")
print(rf_search.best_params_)

print("\nBest CV MAE:")
print(-rf_search.best_score_)
```

```python
best_rf = rf_search.best_estimator_

rf_tuned_train_pred = best_rf.predict(X_train)
rf_tuned_val_pred = best_rf.predict(X_val)
```
